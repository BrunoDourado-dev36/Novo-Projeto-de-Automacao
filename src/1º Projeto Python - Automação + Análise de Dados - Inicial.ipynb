# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# PASSO A PASSO DO DESAFIO 

# 1 - Entrar no sitema da empresa
# 2 - Navegar no sistema para encontrar a base de dados
# 3 - Exportar a base de dados
# 4 - Calcular os indicadores (Faturamento e quantidade de produtos vendidos)
# 5 - Enviar as informacoes por e-mail
```

```python
# 1 - Entrar no sitema da empresa

import pyautogui
import time

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

pyautogui.PAUSE = 1

pyautogui.press("win")
pyautogui.write("chrome")
pyautogui.press("enter")
time.sleep(2)

# Digita o link do sistema
pyautogui.write(link)
pyautogui.press("enter")
time.sleep(2)
```

```python
# 2 - Navegar no sistema para encontrar a base de dados
pyautogui.click(x=375, y=441, clicks=2)
time.sleep(1)

# 3 - Exportar a base de dados
pyautogui.click(x=533, y=444) # seleciona o arquivo
pyautogui.click(x=1817, y=439) # clica nos 3 pontos
pyautogui.click(x=1558, y=530)
time.sleep(3)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas

# 4 - Calcular os indicadores (Faturamento e quantidade de produtos vendidos)

# abrir a base de dados
location = r"C:\Users\Bruno Dourado\Downloads\Vendas - Dez.xlsx"
table = pandas.read_excel(location)

# olhar as informacoes da tabela
display(table)

# somar o faturamento de todos os produtos = somar a coluna de Valor final
faturamento = table["Valor Final"].sum()

# somar a quantidade de produtos = somar a coluna de quantidade
qtd_produtos = table["Quantidade"].sum()

print(faturamento)
print(qtd_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
import pyperclip
# 5 - Enviar as informacoes por e-mail
# abrir uma nova aba
pyautogui.hotkey("ctrl", "t")

# entra no e-mail
pyautogui.write("https://mail.google.com/")
pyautogui.press("enter")
time.sleep(5)
#clica no botao para escrever o e-mail
pyautogui.click(x=103, y=211)
time.sleep(5)

# enviar o e-mail
pyautogui.write("SEU EMAIL AQUI")
pyautogui.press("tab") # seleciona um e-mail
pyautogui.press("tab") # passa para o campo do assunto 

# escreve o assunto
pyperclip.copy("Relatorio de Vendas")
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

# escreve o corpo do e-mail

texto = f"""
Prezados, bom dia!

Segue o relatorio de vendas de hoje.

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: R${qtd_produtos:,}

Antenciosamente,
Bruno Dourado Python
"""
pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")

#envia o e-mail
time.sleep(1)
pyautogui.click(x=1295, y=1002)
```
